```python
"""
CRO Ireland Financial Statements Loader
========================================
Source: https://opendata.cro.ie
Licence: CC BY 4.0

Downloads structured financial data for Irish companies.
We use this as our proxy for unlisted SME balance sheets.
"""
import requests
import pandas as pd
from pathlib import Path

RAW_PATH = Path("data/raw")
RAW_PATH.mkdir(parents=True, exist_ok=True)

CRO_BASE_URL = "https://opendata.cro.ie/api/3/action"


def search_companies(sector_keyword: str, limit: int = 100) -> pd.DataFrame:
    """
    Search CRO dataset for companies by keyword.
    Returns a DataFrame with company numbers and names.
    """
    url = f"{CRO_BASE_URL}/datastore_search"
    params = {
        "resource_id": "d41c8c3a-943a-4b6d-a9a4-f5c2e0b6e5a1",
        "q": sector_keyword,
        "limit": limit
    }
    response = requests.get(url, params=params, timeout=30)
    response.raise_for_status()
    data = response.json()
    return pd.DataFrame(data["result"]["records"])


def get_available_datasets() -> list:
    """
    List all available datasets on CRO open data portal.
    Run this first to find the correct resource IDs.
    """
    url = f"{CRO_BASE_URL}/package_list"
    response = requests.get(url, timeout=30)
    response.raise_for_status()
    return response.json()["result"]


if __name__ == "__main__":
    print("Fetching available CRO datasets...")
    datasets = get_available_datasets()
    for d in datasets:
        print(" -", d)
```
